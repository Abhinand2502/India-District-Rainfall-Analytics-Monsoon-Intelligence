```python
df["Calendar Day"].head(10)
```

```python
date_test = pd.to_datetime(
    df["Calendar Day"],
    errors="coerce"
)
```

```python
print("Minimum date:", date_test.min())
print("Maximum date:", date_test.max())
```

```python
invalid_dates = date_test.isna().sum()

print("Invalid dates:", invalid_dates)
```

```python
date_coverage = (
    pd.DataFrame({
        "date": date_test,
        "year": df["year_num"]
    })
    .dropna(subset=["date"])
    .groupby("year")["date"]
    .nunique()
    .reset_index(name="unique_dates")
)

date_coverage
```

```python
plt.figure(figsize=(9, 5))

plt.bar(
    date_coverage["year"].astype(str),
    date_coverage["unique_dates"]
)

plt.axhline(
    365,
    linestyle="--",
    label="365-day reference"
)

plt.xlabel("Year")
plt.ylabel("Unique dates")
plt.title("Date Coverage by Year")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
top_rainfall = (
    df.nlargest(
        20,
        actual_col
    )
    [[
        "Calendar Day",
        "State",
        "District",
        actual_col,
        normal_col,
        departure_col
    ]]
)

top_rainfall
```

```python
rainfall = df[actual_col]

zero_rain = (rainfall == 0).sum()
nonzero_rain = (rainfall > 0).sum()
missing_rain = rainfall.isna().sum()

print("Zero rainfall:", zero_rain)
print("Positive rainfall:", nonzero_rain)
print("Missing rainfall:", missing_rain)
```

```python
print(
    "Zero rainfall %:",
    (rainfall == 0).mean() * 100
)

print(
    "Positive rainfall %:",
    (rainfall > 0).mean() * 100
)
```

```python
negative_rainfall = (
    df[actual_col] < 0
).sum()

print(
    "Negative rainfall observations:",
    negative_rainfall
)
```
